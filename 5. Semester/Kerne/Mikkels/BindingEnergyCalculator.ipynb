```python
import numpy as np
import astropy.constants as c
import astropy.units as u
import matplotlib.pyplot as plt
```

```python
def BindingEnergy(A,Z):
    N = A - Z
    B = 15.2*u.MeV*A - 15.8*u.MeV*A**(2/3) - 44.7*u.MeV*(N-Z)**2/(2*A) - 0.686*u.MeV*Z**2/(A**(1/3)) \
    + 5*u.MeV*( (-1)**N + (-1)**Z)/(np.sqrt(A))
    return B.to(u.MeV)

def BindingEnergyPerNucleus(A,Z):
    return BindingEnergy(A,Z)/A

def Mass(A,Z):
    m = (Z * (c.m_p + c.m_e) + (A-Z) * c.m_n)*c.c**2  - BindingEnergy(A,Z)
    return m.to(u.MeV)
```
