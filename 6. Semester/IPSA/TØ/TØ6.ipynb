## Opg 6.1
Write a method histogram that given a list of values, returns a list of pairs (value, frequency).

Example. histogram(['A', 'B', 'A', 'A', 'C', 'E', 'C']) should return [('A', 3), ('B', 1), ('C', 2), ('E', 1)].

Hint. Use a dictionary and the dictionary method get.

Note. In the standard library module collections the Counter method implements the same functionality.

`import collections` 

`histogram = list(collections.Counter(['A', 'B', 'A', 'A', 'C', 'E', 'C']).items())`

```python
def histogram(L):
    freq = {}
    for e in L:
        freq[e] = freq.get(e, 0) + 1
    print(freq)
    return [(e, freq[e]) for e in freq]

print(histogram(['A', 'B', 'A', 'A', 'C', 'E', 'C']))
```

_Hint_. Use the `sorted` function to sort a list of tuples of the form (frequency, word), and try to use list comprehension to create such a list from a frequency dictionary.

```python
import re
txt = open(r'C:\Users\Moth\Documents\Uni\6. Semester\IPSA\gerth\solutions\jupyter\saxo.txt', encoding='utf8').read()
txt = txt.lower()
words = re.split('[^a-z]+', txt)

long_words = [w for w in words if len(w) > 5]
hist = histogram(long_words)
hist.sort(key=lambda x: x[1], reverse=True)

print("Rank Freq. Word")
print("====================")
for i, h in enumerate(hist[:10]):
    print(f"{i:4d} {h[1]:5d} {h[0]}")

```
