## Opg 2.4

```python
x = int(input('x: '))

low = 0
high = x + 1

while True:  # low <= sqrt(x) < high
    if low + 1 == high: # If low + 1 is the same as high we are done 
        break

    mid = (high + low) // 2 # Take the average of low and high floored
    print(low, mid, high)

    if mid * mid <= x:  # If mid is less then or equal to sqrt(x)
        low = mid       # Set low to the new mid 
        print('low = mid')
        continue        # Continue will skip high = mid

    high = mid
    print('high = mid')


print(low)  # low = floor(sqrt(x))
```

```python
x = int(input('x: '))

low = 0
high = x + 1

while low + 1 != high:  # Gets rid of break statement

    mid = (high + low) // 2 # Take the average of low and high floored
    print(low, mid, high)

    if mid * mid <= x:  # If mid is less then or equal to sqrt(x)
        low = mid       # Set low to the new mid 
        print('low = mid')
    
    else: # Only executes this line when if does not run
        high = mid
        print('high = mid')


print(f'Result: {low}')  # low = floor(sqrt(x))
```

## Opg 2.5

```python
x = int(input("Number: "))
f = 2

while f != x:
    if x % f == 0:
        print(str(x)+ " is not a prime, factor: " + str(f))
        break
    else: 
        f += 1

if x == f: 
    print(str(x) + " is a prime")
```

```python
for x in range(2,30):
    f = 2

    factors = []

    while f != x:
        if x % f == 0:
            factors.append(f)
            f += 1
        else: 
            f += 1

    if factors == []: 
        print(f'{str(x)} is a prime')
    else: 
        print(f'{str(x)} is not a prime \n Factors: {factors}')
```

```python
import math

n = int(input('n: '))

try: 
    f = math.factorial(n)
    s = (2 * math.pi * n) ** 0.5 * (n/math.e)**n
    print(f'n = {n}, n! = {f:.3g}, S = {s:.3g}, acc = {f/s}')
except Exception as e:
    print(e)

log_f = math.log(f)
log_s = 0.5 * math.log(2*math.pi*n) + n * math.log(n/math.e)
print(f'n = {n}, ln(n)! = {log_f:.3g}, ln(S) = {log_s:.3g}, acc = {log_f/log_s}')
```

```python
import time
# Time_it decorator from lectures, used to time functions. 
def time_it(f):
    def wrapper(*args, **kwargs):
        t_start = time.time()
        result = f(*args, **kwargs)
        t_end = time.time()
        t = t_end - t_start
        print(f'{f.__name__} took {t:.2f} seconds')
        return result
    return wrapper


n = int(input())

@time_it
def app(n): 
    l = []
    for i in range(n):
        l.append(i)

@time_it
def sli(n): 
    l = [None] * n
    for i in range(n):
        l[i] = n

app(n)
sli(n)
```
