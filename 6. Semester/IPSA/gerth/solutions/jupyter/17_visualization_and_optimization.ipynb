## Exercise 17.3 (Jupyter plotting)

Plot the function _f_(_x_) = 100·(_x_ - 3)<sup>2</sup> + _e_<sup>_x_</sup> for the _x_ in ther range [0, 10] using `matplotlib.pyplot` in Jupyter.

```python
%%time
%matplotlib inline
#%matplotlib widget

import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return 100*(x - 3)**2 + np.e**x

x = np.linspace(0, 10, 1000)

plt.semilogy()
plt.plot(x, f(x))
plt.show()
```

## Exercise 17.4 (Jupyter function minimization)

Find the minimum of the function in Exercise 17.3 using `scipy.optimize.minimize`.  Extend the plot in Exercise 17.3 to illustrate the minimum, e.g. using a dot.

```python
from scipy.optimize import minimize

x_min = minimize(f, [0.0]).x[0]

print(x_min, f(x_min))

plt.semilogy()
plt.plot(x, f(x))
plt.plot(x_min, f(x_min), 'ro')
plt.show()

minimize(f, [0.0])
```
