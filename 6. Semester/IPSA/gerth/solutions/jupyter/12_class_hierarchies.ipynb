## Exercise 12.1 (geometric shapes)

Make a class `Shape` that represent an abstract geometric shape.  Concrete shapes, like circles, axis-aligned rectangles, and triangles should be defined as subclasses of `Shape`.  The class `Shape` should only define a method `fatness` that depends on methods `area` and `perimeter` that should be defined in  subclasses:

* `fatness()` return 4·π·area / perimeter<sup>2</sup>.

Subclasses should implement the following methods (but note the method `fatness`):

* `__init__` initialize the geometric object. The arguments to `__init__` will depend on the geometric shape.

* `__str__` return a string describing the object, e.g. `Circle(radius=3, center_x=4, center_y=5)`.

* `area()` return the area of a geometric object.

* `perimeter()` return the length of the perimeter of the geometric object.

* `contains(x, y)` returns if the point (`x`, `y`) is contained within the shape.

Create classes `Circle`, `Rectangle`, and `Triangle`, each inheriting from class `Shape`. Implementing class `Triangle` is optional (the methods `area` and `contains` require some amount of math, with solutions easily to find on the internet).  For each subclass implement the methods `__int__`, `__str__`, `area`, `perimeter`, and `contains`.

It should be possible to create instances of the geometric objects (calling `__init__`) as follows:

* `Circle(radius, center_x, center_y)`

* `Rectangle(min_x, max_x, min_y, max_y)`

* `Triangle(x0, y0, x1, y1, x2, y2)`

_Example_.

    > shapes = [Rectangle(0, 10, 0, 1), Circle(4, 1, 4), Triangle(1, 2, 1, 4, 2, 3)]
    > [str(s) for s in shapes]
    ['Rectangle((0, 0), (10, 1))', 'Circle(radius=4, center_x=1, center_y=4)', 'Triangle((1, 2), (1, 4), (2, 3))']
    > [s.area() for s in shapes]

    [22, 25.132741228718345, 4.82842712474619]

    [0.25963575649502424, 1.0, 0.5390120844526473]

```python
from math import pi, sqrt


class Shape:
#    def __init__(self):
#        raise NotImplementedError

    def fatness(self):
        if self.perimeter() == 0:
            return 0
        else:
            return 4 * pi * self.area() / self.perimeter()**2

    # The following are not required, but would be polite to include
    def area():
        # Should be implemented by subclass
        raise NotImplementedError

    def perimeter():
        # Should be implemented by subclass
        raise NotImplementedError

class Circle(Shape):
    def __init__(self, radius=1, center_x=0, center_y=0):
        self.radius = radius
        self.x = center_x
        self.y = center_y

    def area(self):
        return self.radius**2 * pi

    def perimeter(self):
        return 2 * self.radius * pi

    def contains(self, x, y):
        return (self.x - x)**2 + (self.y - y)**2 <= self.radius**2

    def __str__(self):
        return 'Circle(radius=%s, center_x=%s, center_y=%s)' % (self.radius, self.x, self.y)


class Rectangle(Shape):
    def __init__(self, min_x=0, max_x=1, min_y=0, max_y=1):
        self.min_x = min_x
        self.max_x = max_x
        self.min_y = min_y
        self.max_y = max_y

    def area(self):
        return (self.max_x - self.min_x) * (self.max_y - self.min_y)

    def perimeter(self):
        return 2 * (self.max_x - self.min_x) + 2 * (self.max_y - self.min_y)

    def contains(self, x, y):
        return self.min_x <= x <= self.max_x and self.min_y <= y <= self.max_y

    def __str__(self):
        return ('Rectangle((%s, %s), (%s, %s))'
               % (self.min_x, self.min_y, self.max_x, self.max_y))


class Triangle(Shape):
    def __init__(self, x0=0, y0=0, x1=1, y1=0, x2=0.5, y2=1):
        self.x0 = x0
        self.y0 = y0
        self.x1 = x1
        self.y1 = y1
        self.x2 = x2
        self.y2 = y2

    def __str__(self):
        return ('Triangle((%s, %s), (%s, %s), (%s, %s))'
               % (self.x0, self.y0, self.x1, self.y1, self.x2, self.y2))

    def perimeter(self):
        def dist(x0, y0, x1, y1):
            return sqrt((x1 - x0)**2 + (y1 - y0)**2)

        return (dist(self.x0, self.y0, self.x1, self.y1) +
                dist(self.x1, self.y1, self.x2, self.y2) +
                dist(self.x0, self.y0, self.x2, self.y2))

    # https://en.wikipedia.org/wiki/Shoelace_formula
    def area(self):
        return abs((self.x0-self.x2)*(self.y1-self.y0)
                  -(self.x0-self.x1)*(self.y2-self.y0))/2

    # https://stackoverflow.com/questions/2049582/how-to-determine-if-a-point-is-in-a-2d-triangle
    def contains(self, x, y):
        def sided(p1, p2, p3):
            return (p1[0] - p3[0]) * (p2[1] - p3[1]) - (p2[0] - p3[0]) * (p1[1] - p3[1])

        pt = (x,y)
        v1 = (self.x0, self.y0)
        v2 = (self.x1, self.y1)
        v3 = (self.x2, self.y2)

        b1 = sided(pt, v1, v2) <= 0.0
        b2 = sided(pt, v2, v3) <= 0.0
        b3 = sided(pt, v3, v1) <= 0.0

        return b1 == b2 == b3


shapes = [Rectangle(0, 10, 0, 1), Circle(4, 1, 4), Triangle(1, 2, 1, 4, 2, 3)]
print([str(s) for s in shapes])
print([s.area() for s in shapes])
print([s.perimeter() for s in shapes])
print([s.fatness() for s in shapes])
print([s.contains(1, 1) for s in shapes])

triangle1 = Triangle(1, 2, 1, 4, 2, 3)
triangle2 = Triangle(1, 2, 2, 3, 1, 4)
print(triangle1.contains(1, 3))
print(triangle2.contains(1, 3))
```

## Exercise 12.2 - handin 6 (ancestors)

Create a class `Person` to represent a person with a name, year of birth, year of death (`None` if the person is still alive), and with references to two other `Person` objects representing the father and mother of the person. The class should define the following three methods:

* A constructor `Person(name, mother, father, born, died)` where all arguments have default value `None`.

* `__str__` returning the string `'name born-died'`.

* `ancestors()` returns a tuple `(name-string, recursive-father-ancestors, recursive-mother-ancestors)` (see example below).

Create a second class `AnnotatedPerson` that is a subclass of `Person`, that allows each person to be annotated with a `note`, that can be given as an argument to the constructor, and will be part of the string returned by `__str__`, i.e. redefine the methods `__init__` and `__str__`.  The method `ancestors` should be inherited from `Person` and not be redefined in `AnnotatedPerson`. Try to avoid repeating code from the super class, instead call the methods in the super class.

Using the two classes you should be able to create the following objects:

    louise_af_HK = Person('Louise of Hessen-Kassel', None, None, 1817, 1898)
    christian_9  = Person('Christian 9.', None, None, 1818, 1906)
    louise_af_SN = AnnotatedPerson('Louise of Sweden-Norway', None, None, 1851, 1926, 'born Princess of Sweden and Norway')
    frederik_8   = Person('Frederik 8.', louise_af_HK, christian_9, 1870, 1947)
    christian_10 = Person('Christian 10.', louise_af_SN, frederik_8, 1870, 1947)
    ingrid       = AnnotatedPerson('Ingrid of Sweden', None, None, 1910, 2000, 'Queen of Denmark 1947-1970')
    frederik_9   = Person('Fredrik 9.', None, christian_10, 1899, 1972)
    margrethe_ii  = AnnotatedPerson('Margrethe II', ingrid, frederik_9, 1940, note='Queen of Denmark')`

Calling `margrethe_ii.ancestors()` should return something like: `('Margrethe II 1940- [Queen of Denmark]', ('Fredrik 9. 1899-1972', ('Christian 10. 1870-1947', ('Frederik 8. 1870-1947', ('Christian 9. 1818-1906', '-', '-'), ('Louise of Hessen-Kassel 1817-1898', '-', '-')), ('Louise of Sweden-Norway 1851-1926 [born Princess of Sweden and Norway]', '-', '-')), '-'), ('Ingrid of Sweden 1910-2000 [Queen of Denmark 1947-1970]', '-', '-'))`

Applying the tree printing program of Exercise 9.2 to this output would print something like the below (creating the recursive print out is not part of the exercise).

    --Margrethe II 1940- [Queen of Denmark]
      |--Fredrik 9. 1899-1972
      |  |--Christian 10. 1870-1947
      |  |  |--Frederik 8. 1870-1947
      |  |  |  |--Christian 9. 1818-1906
      |  |  |  |  |---
      |  |  |  |  |---
      |  |  |  |--Louise of Hessen-Kassel 1817-1898
      |  |  |  |  |---
      |  |  |  |  |---

      |  |  |  |---
      |  |  |  |---
      |  |---
      |--Ingrid of Sweden 1910-2000 [Queen of Denmark 1947-1970]
      |  |---
      |  |---

```python
class Person:
    def __init__(self,
                 name=None,
                 mother=None,
                 father=None,
                 born=None,
                 died=None):
        self.name = name
        self.mother = mother
        self.father = father
        self.born = born
        self.died = died

    def __str__(self):
        txt = self.name
        if self.born != None or self.died != None:
            txt += ' '
            if self.born != None:
                txt += str(self.born)
            txt += '-'
            if self.died != None:
                txt += str(self.died)
        return txt

    def ancestors(self):
        father_side = self.father.ancestors() if self.father != None else '-'
        mother_side = self.mother.ancestors() if self.mother != None else '-'
        return (str(self), father_side, mother_side)


class AnnotatedPerson(Person):
    def __init__(self,
                 name=None,
                 mother=None,
                 father=None,
                 born=None,
                 died=None,
                 note=None):
        self.note = note
        #super().__init__(name, mother, father, born, died)
        Person.__init__(self, name, mother, father, born, died)


    def __str__(self):
        # txt = super().__str__()
        txt = Person.__str__(self)
        if self.note != None:
            txt += ' [' + self.note +']'
        return txt

# Exercise 9.2
def print_tree(tree, prefix=''):
    label, *children = tree
    print(prefix + '--' + label)
    for c in children:
        print_tree(c, prefix=prefix + '  |')


louise_af_HK = Person('Louise af Hessen-Kassel', None, None, 1817, 1898)
christian_9  = Person('Christian 9.', None, None, 1818, 1906)

louise_af_SN = AnnotatedPerson('Louise of Sweden-Norway', None, None, 1851, 1926, 'født Prinsesse af Sverige')
frederik_8   = Person('Frederik 8.', louise_af_HK, christian_9, 1870, 1947)

christian_10 = Person('Christian 10.', louise_af_SN, frederik_8, 1870, 1947)

ingrid       = AnnotatedPerson('Ingrid of Sweden', None, None, 1910, 2000, 'Queen of Denmark 1947-1970')
frederik_9   = Person('Fredrik 9.', None, christian_10, 1899, 1972)

magrethe_ii  = AnnotatedPerson('Magrethe II', ingrid, frederik_9, 1940, note='Queen of Denmark')

print(magrethe_ii.ancestors())
print_tree(magrethe_ii.ancestors())
```
