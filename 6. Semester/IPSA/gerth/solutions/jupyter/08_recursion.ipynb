## Exercise 8.1 (upper and lower cases)

Make a recursive function `cases(s)` that given a string `s`, generates a list of all possible upper and lower case combinations of the letters in the string. E.g. `cases('abcB')` should return a list containing the following 16 strings `['abCb', 'abCB', 'abcb', 'abcB', 'aBCb', 'aBCB', 'aBcb', 'aBcB', 'AbCb', 'AbCB', 'Abcb', 'AbcB', 'ABCb', 'ABCB', 'ABcb', 'ABcB']`, not necessarily in this order.

_Hint_. Use `.lower()` and `.upper()` to convert a string containing one letter to lower and upper case.

```python
# Later in the course, the students are introduced to 'decorators',
# where the below decorator 'trace' is useful for understanding recursion.
# Can be found in all-slides.pdf by searching for '@trace'.

def trace(f):  # decorator to trace recursive calls
    indent = 0
    def wrapper(*args):
        nonlocal indent
        spaces = '| ' * indent
        arg_str = ', '.join(map(repr, args))
        print(spaces + f'{f.__name__}({arg_str})')
        indent += 1
        result = f(*args)
        indent -= 1
        print(spaces + f'> {result}')
        return result
    return wrapper

@trace
def binomial(n, k):
    if k == 0 or k == n:
        return 1
    return binomial(n - 1, k) + binomial(n - 1, k - 1)


print(binomial(3, 2))
```

```python
# @trace
def cases(s):
    if not s:
        return [ '' ]

    first = s[0]
    tail = s[1:]
    tails = cases(tail)
    firsts = sorted({first.lower(), first.upper()})
    return [c + tail for tail in tails for c in firsts]


print(cases('abc'))
print(cases('foo_bar'))
```

```python
def cases(s):
    if not s:
        return [ '' ]

    first = s[0]
    tails = cases(s[1:])
    answer = [first.lower() + tail for tail in tails]
    if first.lower() != first.upper():
        answer += [first.upper() + tail for tail in tails]
    return answer

if __name__ == '__main__':
    print(cases('foo_bar'))
```

## Exercise 8.2 (list subsets)

Make a recursive function `subsets(L)` that given a list `L` returns a list of all subsets of `L` (each subset being a list). E.g. `subsets([1, 2])` should return `[[], [1], [2], [1, 2]]`. The order of the returned lists can be arbitrary.

```python
'''
Given a list L, returns a list of all subsets of L as lists

Example: subsets([1, 2]) -> [[], [1], [2], [1, 2]]
'''

def subsets(S):
    if not S:
        return [[]]
    T = subsets(S[1:])
    return T + [[S[0]] + t for t in T]
#    return T + [[S[0], *t] for t in T]


print(subsets([1, 2, 3]))
```

```python
def subsets(S):
    return [y + x for x in subsets(S[1:]) for y in ([], [S[0]])] if S else [[]]

print(subsets([1, 2, 3]))
```

## Exercise 8.3 (binary strings)

Create a recursive function `binary(n, k)` that returns a list of all binary strings of length `n` and containing `k` ones (0 &le; `k` &le; `n`).

E.g. `binary(5, 3)` should return the list `['00111', '01011', '01101', '01110', '10011', '10101', '10110', '11001', '11010', '11100']` (in any order).

```python
def binary(n, k):
    '''List of all binary strings of length n with k ones.'''

    if k == 0:
        return ['0' * n]
    if k == n:
        return ['1' * n]
    return ['0' + s for s in binary(n - 1, k)] + ['1' + s for s in binary(n - 1, k - 1)]

def binary(n, k):
    '''List of all binary strings of length n with k ones.'''

    if k == 0:
        return ['0' * n]
    return ['0' * i + '1' + s for i in range(n - k + 1) for s in binary(n - i - 1, k - 1)]

def binary(n, k, suffix=''):
    '''List of all binary strings of length n with k ones appended with suffix.'''

    if k == 0:
        return ['0' * n + suffix]
    if k == n:
        return ['1' * n + suffix]
    return binary(n - 1, k, '0' + suffix) + binary(n - 1, k - 1, '1' + suffix)

print(f'{binary(5, 3) = }')

# Test correctness on a larger example

n = 20
k = 10
B = binary(n, k)
print(f'{len(B)}')

import math
assert len(B) == len(set(B))  # all output distinct
assert all([len(b) == n for b in B])
assert all([b.count('1') == k for b in B])
assert all([b.count('0') + b.count('1') == n for b in B])
assert len(B) == math.comb(n, k)  # the number of strings is given by the binomial coefficient
```

## Exercise 8.4 (tree relabeling)

Make a recursive function `relabel(tree, new_names)` that takes a tree `tree` and a dictionary `new_names = {old_name: new_name, ...}`, and returns a new tree where labels in the dictionary `new_names` are replaced by the corresponding values in the dictionary. Leaves not in the dictionary remain unchanged.

_Example_. `relabel(('a', ('b', 'c')), {'a': 'x', 'c': 'y'})` should return `('x', ('b', 'y'))`.

_Hint_. Start making `relabel` for binary trees only. Use the method `isinstance` to check if a value is of class `tuple` or a `str`.

```python
def relabel(tree, new_names):
    if isinstance(tree, str):
        # return new_names[tree] if tree in new_names else tree
        return new_names.get(tree, tree)
    return tuple([relabel(child, new_names) for child in tree])

    # if binary tree, one could do the following
    left, right = tree
    return (relabel(left, new_names), relabel(right, new_names))

tree = ('a', ('b', 'c'))
new_names = {'a': 'x', 'c': 'y'}

print(tree, '->', relabel(tree, new_names))
```

## Exercise 8.5* (validate leaf-labeled binary trees)

Assume we want to represent binary trees, where each leaf has a string as a label, by nested tuples. We require the leaves are labeled with distinct non-empty strings and all non-leaf nodes have exactly two children. E.g. `((('A', 'B'), 'C'), ('D', ('F', 'E')))` is a valid binary tree.

1.  Write a function `validate_string_tuple(t)` that checks, i.e. returns `True` or `False`, if the value `t` is a tuple only containing distinct strings, e.g. `('a', 'b', 'c')`.


2.  Write a recursive function `valid_binary_tree(tree)` program that checks, i.e. returns `True` or `False`, if the value `tree` is a recursive tuple representing a binary tree as described above.


_Hint_. Use the method `isinstance` to check if a value is of class `tuple` or a `str`, and use a recursive function to traverse a tree. Collect all leaf labels in a list, and check if all leaves are distinct by converting to `set`.

```python
'''
Verify that a value represents a binary tree by a recursive tuple where
the leaves are distinct non-empty strings and all non-leaf nodes
have two children.

E.g. (('A','B'),'C') represents the below tree

     /\
    /  'C'
   /\
'A'  'B'
'''


def validate_string_tuple(t):
    return (isinstance(t, tuple)
            and all([isinstance(c, str) for c in t])
            and len(t) == len(set(t)))


def validate_tree(tree):
    ''' validate tree according to input specification '''

    labels = set()  # set of labels seen

    def recursive_validate(tree):
        ''' recursive traversal of tree
            visited leaf labels are in the labels set '''

        nonlocal labels  # also works without nonlocal

        if isinstance(tree, str) and tree != '' and tree not in labels:
            labels.add(tree)
            return True
        if isinstance(tree, tuple) and len(tree) == 2:
            return all([recursive_validate(child) for child in tree])
        return False

    return recursive_validate(tree)


for t in (('a', 'b', 'c'),
          ('a', 'b', 'b'),
          ('a', 1, 'c'),
          42):
    print(t, validate_string_tuple(t))

trees = [
    ((('A', 'B'), 'C'), ('D', ('F', 'E'))),
    ((('A', 'F'), 'B', 'D'), ('C', ('E', ()))),
    ((('A', 'F'), 'B'), 'A'),
    ('A', 'B', 'C'),
    (('A', ''), 'C')
]

for t in trees:
    print(validate_tree(t), t)
```

## Exercise 8.6* (subtree extraction)

Make a recursive function `extract(tree, leaves)` that takes a binary tree `tree` and a set of `leaves`, generates a new binary tree only containing the leaves in `leaves`, i.e. repeatedly all other leaves are pruned, previous internal nodes with no children are removed, and previous internal nodes with only one child are replaced by the child.

_Example_. `extract(((('a', 'b'), 'c'), ((('d', 'e'), 'f'), 'g')), {'a', 'c', 'd', 'f', 'g'})` should return `(('a', 'c'), (('d', 'f'), 'g'))` (see illustration below where leaves `'b'` and `'e'` have been removed).

               before                   after
                 /\                       /\
                /  \                     /  \
               /    \                   /    \
              /      \                 /\     \
             /        \               /  \     \
            /          \            'a'  'c'   /\
           /            \                     /  \
          /\            /\                   /    \
         /  \          /  \                 /\    'g'
        /    \        /    \               /  \
       /\    'c'     /\     \            'd'  'f'
      /  \          /  \    'g'
    'a'  'b'       /    \
                  /\    'f'
                 /  \
               'd'  'e'

```python
def extract(tree, leaves):
    if isinstance(tree, str):
        return tree if tree in leaves else None
    children = [extract(child, leaves) for child in tree]
    children = [child for child in children if child != None]
    if len(children) == 0:
        return None
    if len(children) == 1:
        return children[0]
    return tuple(children)

def extract(tree, leaves):
    if isinstance(tree, str):
        return tree if tree in leaves else None
    children = [extracted for child in tree if (extracted := extract(child, leaves))]
    if len(children) == 0:
        return None
    if len(children) == 1:
        return children[0]
    return tuple(children)

tree = ((('a', 'b'), 'c'), ((('d', 'e'), 'f'), 'g'))

new_tree = extract(tree, {'a', 'c', 'd', 'f', 'g'})

print(tree, '->', new_tree)
```
