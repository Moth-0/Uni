Assume you have _n_ jobs where each job needs to be performed on some machine at a particular interval of time, i.e. the _i_-th task has a start time _start_<sub>_i_</sub> and finishing time _end_<sub>_i_</sub>.  You are going to buy a number of identical machines and to schedule the jobs on the machines such that two jobs scheduled on the same machine do not have overlapping time intervals (not even at the endpoints).

The problem is to _minimize_ the number of machines required to execute all the _n_ jobs.

The below list of eight time intervals for eight jobs can be scheduled on three machines as shown in the figure.

    [(1, 5), (6, 7), (3, 8), (2, 4), (8, 10), (5, 9), (1, 2), (9, 10)]

        [2, 2, 1, 3, 2, 3, 1, 1]


2.  Write a function `visualize(jobs, schedule)` that visualizes the solution using matplotlib.pyplot, e.g. as in the figure.

_Hint_. Sort the jobs by start time and assign the jobs to machines by considering them in increasing order of start time. Use the module [`heapq`](https://docs.python.org/3.0/library/heapq.html) to keep track of the currently active jobs and their finishing times. Whenever the earliest finishing time of an active job is earlier than the start time of the next job to schedule, terminate this active job and release its machine to a pool of free machines.

```python
import heapq
import matplotlib.pyplot as plt


def solve(jobs):
    schedule = [None] * len(jobs)

    tasks = sorted([(start, end, idx) for idx, (start, end) in enumerate(jobs)])

    number_of_machines = 0
    working = []  # heapq of (end, idx)
    free_machines = []  # list of machine IDs

    for start, end, idx in tasks:
        while working and working[0][0] < start:
            finish, machine = heapq.heappop(working)
            free_machines.append(machine)

        if not free_machines:
            number_of_machines += 1
            free_machines.append(number_of_machines)

        machine = free_machines.pop()
        schedule[idx] = machine
        heapq.heappush(working, (end, machine))

    return schedule


def solve(jobs):
    schedule = [None] * len(jobs)

    tasks = sorted([(start, end, idx) for idx, (start, end) in enumerate(jobs)])

    machines = []  # heapq of (end, idx)

    for start, end, idx in tasks:
        if not machines or machines[0][0] >= start:
            machine = len(machines) + 1
        else:
            end_machine, machine = heapq.heappop(machines)

        schedule[idx] = machine
        heapq.heappush(machines, (end, machine))

    return schedule


def visualize(jobs, schedule):
    for idx, ((start, end), machine) in enumerate(zip(jobs, schedule)):
        plt.plot([start, end], [machine, machine], 'go-')
        plt.text((start + end) / 2, machine + 0.1, str(idx),
                 horizontalalignment='center')

    plt.ylabel('Machine')
    plt.xlabel('Time')
    plt.ylim(0, max(schedule) + 1)
    plt.yticks(range(1, max(schedule) + 1))
    plt.show()


if __name__ == '__main__':
    jobs = [(1, 5),
            (6, 7),
            (3, 8),
            (2, 4),
            (8, 10),
            (5, 9),
            (1, 2),
            (9, 10)]

    schedule = solve(jobs)
    print(schedule)
    visualize(jobs, schedule)

    # from job_scheduling import solve
    from random import randint
    from time import time

    jobs = [(i + randint(0, 100), i + randint(100, 200)) for i in range(1000000)]

    start = time()
    schedule = solve(jobs)
    end = time()
    print(max(schedule), 'machines')
    print('Running time:', end - start)
```
