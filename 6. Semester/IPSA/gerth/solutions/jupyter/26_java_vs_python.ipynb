## Exercise 26.3 (down payment in Java)

Implement Exercise 2.8 (down payment) in Java.

To read a double (64 bit floating point value) from the input you can use the following statement:

    double loan = Double.parseDouble(System.console().readLine());

or using the `java.util.Scanner` class:

    Scanner sc = new Scanner(System.in);
    double loan = sc.nextDouble();

```python
// File: DownPayment.java

import java.time.Month;
import java.util.Scanner;

public class DownPayment {
    static double max(double x, double y) {
        return x < y ? y : x;
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("Initial loan:");
        double loan = sc.nextDouble();

        System.out.println("Monthly interest rate %:");
        double rate = sc.nextDouble();

        System.out.println("Monthly payment:");
        double monthly_payment = sc.nextDouble();

        System.out.println("months and remaining loan");
        int months = 0;
        while (loan > 0) {
            double interest = loan * rate / 100;
            loan = max(0, loan + interest - monthly_payment);
//          loan = Double.max(0, loan + interest - monthly_payment);
            months ++;
            System.out.println(months + " " + loan);
        }
    }
}
```
