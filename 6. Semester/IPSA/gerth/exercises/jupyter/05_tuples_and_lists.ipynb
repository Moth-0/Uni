## Exercise 5.3 (zip)

Implement your own version `my_zip` of `zip`, such that
`my_zip([L1, L2,..., Lk]) == list(zip(L1, L2, ...., Lk))`.

```python
L1 = [1,2,3]
L2 = ["a","b","c"]
L3 = [True,True,False]

def my_zip(LL):
    n = min([len(l) for l in LL])
    return [tuple(L[i] for L in LL) for i in range(n)]

print(list(zip(L1, L2, L3)))    
print(my_zip([L1,L2,L3]))
```
