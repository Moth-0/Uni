## Exercise 17.3 (Jupyter plotting)

Plot the function _f_(_x_) = 100·(_x_ - 3)<sup>2</sup> + _e_<sup>_x_</sup> for the _x_ in ther range [0, 10] using `matplotlib.pyplot` in Jupyter.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

def f(x):
    return 100*(x-3)**2+np.exp(x)

x = np.linspace(0, 10, 1000)

f_min = minimize(f, 2)

print(f_min)

plt.plot(x, f(x))
plt.plot(f_min.x, f_min.fun, "o")
plt.yscale("log")
plt.show()
```
