## Exercise 2.6 (integer square root)

Consider the following program for finding the integer square root of a non-negative integer:

    x = int(input('x: '))

    low = 0
    high = x + 1

    while True:  # low <= sqrt(x) < high
        if low + 1 == high:
            break

        mid = (high + low) // 2

        if mid * mid <= x:
            low = mid
            continue

        high = mid

    print(low)  # low = floor(sqrt(x))

## Exercise 2.8 (integer factorization)

Recall that any positive number has a unique prime factorization, e.g. 126 = 2 * 3 * 3 * 7.

The program should ask the user for the integer _x_, and then print each prime factor.
Note that for _x_ = 126, the prime factor 3 should be printed two times.

_Hint_. Repeatedly divide _x_ by 2 until 2 is not a prime factor any longer. Repeat the same for 3, 4, 5,... Note that 4 = 2 * 2 is not a prime number.
