## Exercise 8.1 (upper and lower cases)

Make a recursive function `cases(s)` that given a string `s`, generates a list of all possible upper and lower case combinations of the letters in the string. E.g., `cases('abcB')` should return a list containing the following 16 strings `['abCb', 'abCB', 'abcb', 'abcB', 'aBCb', 'aBCB', 'aBcb', 'aBcB', 'AbCb', 'AbCB', 'Abcb', 'AbcB', 'ABCb', 'ABCB', 'ABcb', 'ABcB']`, not necessarily in this order.

_Hint_. Use `.lower()` and `.upper()` to convert a string containing one letter to lower and upper case.

## Exercise 8.2 (list subsets)

Make a recursive function `subsets(L)` that given a list `L` returns a list of all subsets of `L` (each subset being a list). E.g., `subsets([1, 2])` should return `[[], [1], [2], [1, 2]]`. The order of the returned lists can be arbitrary.

## Exercise 8.3 (binary strings)

E.g., `binary(5, 3)` should return the list `['00111', '01011', '01101', '01110', '10011', '10101', '10110', '11001', '11010', '11100']` (in any order).

## Exercise 8.4 (label function)

Assume you enumerate all nonempty strings with letters A-Z as follows:

    A,B,...,Z,AA,AB,...,AZ,BA,...,BZ,...,ZZ,AAA,AAB,...,ZZZ,AAAA,...

I.e., the strings are sorted first by length and then lexicographically.
We number the strings 0,1,2,... such that the first 26 strings are the letters
`A` to `Z`, the next 26 ** 2 strings the strings `AA` to `ZZ` with two letters
and which have numbers 26,...,26+26**26-1, ...

Create a function `label` that takes as argument a single non-negative integer
`i` and prints the `i`th label in the enumeration above.

_Example_:

    for i in [0, 1, 2, 3, 24, 25, 26, 27, 51, 52, 26+26*26-1, 26+26*26, 201883747]:
        print(i, label(i))

should print

    0 A
    1 B
    2 C
    3 D
    24 Y
    25 Z
    26 AA
    27 AB
    51 AZ
    52 BA
    701 ZZ
    702 AAA
    201883747 PYTHON

_Hint_. Think recursively. If `i` is less than 26, the string only
consists of a single letter. Otherwise, the string consists of the last letter
and a prefix (a label with one letter less) that changes whenever `i` increases by 26.

## Exercise 8.5 (tree relabeling)

Make a recursive function `relabel(tree, new_names)` that takes a tree `tree` and a dictionary `new_names = {old_name: new_name, ...}`, and returns a new tree where labels in the dictionary `new_names` are replaced by the corresponding values in the dictionary. Leaves not in the dictionary remain unchanged.

_Example_. `relabel(('a', ('b', 'c')), {'a': 'x', 'c': 'y'})` should return `('x', ('b', 'y'))`.

_Hint_. Start making `relabel` for binary trees only. Use the method `isinstance` to check if a value is of class `tuple` or a `str`.

1.  Write a function `validate_string_tuple(t)` that checks, i.e., returns `True` or `False`, if the value `t` is a tuple only containing distinct strings, e.g., `('a', 'b', 'c')`.
