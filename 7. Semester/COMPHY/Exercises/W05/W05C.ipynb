```python
import numpy as np
import matplotlib.pyplot as plt

def P1(x):
    x = np.asarray(x)
    out = np.zeros_like(x)
    mask = (abs(x) <= 2)
    out[mask] = 1.0/(np.pi*np.sqrt(4 - x[mask]**2))
    return out

def P2(x, kbT=2):
    return 1.0/np.sqrt(2*np.pi*kbT) * np.exp(-x**2/(2*kbT))

fig, ax = plt.subplots(1,2, figsize=(10, 5))

x = np.linspace(-3, 3, 1000)
ax[0].plot(x, P1(x), label='P1')
ax[1].plot(x, P2(x), label='P2')
ax[0].set_xlabel('x')
ax[0].set_ylabel('P1(x)')
ax[1].set_xlabel('x')
ax[1].set_ylabel('P2(x)')
plt.tight_layout()
```

```python
from scipy.integrate import quad

int_p1 = quad(P1, -2, 2)
int_p2 = quad(P2, -np.inf, np.inf)

print(f'Integraal of P1 = {int_p1[0]:.3f}')
print(f'Integraal of P2 = {int_p2[0]:.3f}')
```

```python
avg_V1 = lambda x: P1(x)*x**2/2.0
avg_V2 = lambda x: P2(x)*x**2/2.0

int_V1 = quad(avg_V1, -2, 2)
int_V2 = quad(avg_V2, -np.inf, np.inf)

print(f'<V>1 = {int_V1[0]:.3f}')
print(f'<V>2 = {int_V2[0]:.3f}')
```

```python
s1 = lambda x: -P1(x)*np.log(P1(x))
s2 = lambda x: -P2(x)*np.log(P2(x))

S1 = quad(s1, -2, 2)
S2 = quad(s2, -50, 50)

print(f'S1/kb = {S1[0]:.3f} = ln(π)')
print(f'S2/kB = {S2[0]:.3f} = 1/2 ln(4π) + 1/2')
```

## The distribution would still integrate to 1, and both entropys would get the same factor added on, giving us the same difference between them.
