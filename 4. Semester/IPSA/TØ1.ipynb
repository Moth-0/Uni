```python
k = int(input("How many numbers: "))
sum = 0
while k > 0:
    x = int(input("Number to add: "))
    sum += x 
    k -= 1

print(sum)
```

```python
a = int(input("a: "))
b = int(input("b: "))
c = int(input("c: "))
d = int(input("d: "))

ab = [a,b]
cd = [c,d]

if b >= c and d >= a:
    over = [max(a,c), max(b,c)]
    print(str(ab) + " and " + str(cd) + " overlap in the interval " + str(over))
else:
    print(str(ab) + " and " + str(cd) + " do not overlap ")
```

```python
x = int(input('x: '))

low = 0
high = x + 1

while True:  # low <= sqrt(x) < high
    if low + 1 == high:
        break

    mid = (high + low) // 2

    if mid * mid <= x:
        low = mid
        continue

    high = mid

print(low)  # low = floor(sqrt(x))
```

```python
x = int(input("Number: "))
f = 2

while f != x:
    if x % f == 0:
        print(str(x)+ " is not a prime, factor: " + str(f))
        break
    else: 
        f += 1

if x == f: 
    print(str(x) + " is a prime")
```

```python
for x in range(2,30):
    f = 2

    while f != x:
        if x % f == 0:
            print(str(x)+ " is not a prime, factor: " + str(f))
            break
        else: 
            f += 1

    if x == f: 
        print(str(x) + " is a prime")
```
